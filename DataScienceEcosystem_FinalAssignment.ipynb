### Câu 10 & 11: Phân tích biểu đồ hộp cho biến MEDV và theo nhóm tuổi (AGE)
Tạo nhóm tuổi cho biến AGE (<=35, 35-70, >=70) và vẽ Boxplot so sánh giá nhà MEDV.

```python
boston_df.loc[boston_df['AGE'] <= 35, 'age_group'] = '35 tuổi trở xuống'
boston_df.loc[(boston_df['AGE'] > 35)&(boston_df['AGE'] < 70), 'age_group'] = 'Từ 35 đến 70 tuổi'
boston_df.loc[boston_df['AGE'] >= 70, 'age_group'] = '70 tuổi trở lên'

plt.figure(figsize=(8,5))
sns.boxplot(x='age_group', y='MEDV', data=boston_df, order=['35 tuổi trở xuống', 'Từ 35 đến 70 tuổi', '70 tuổi trở lên'])
plt.title('Giá nhà (MEDV) theo nhóm tuổi của căn nhà (AGE)')
plt.xlabel('Nhóm tuổi nhà')
plt.ylabel('Giá trị trung vị căn nhà ($1000s)')
plt.show()
```

**Nhận xét (Câu 10 & 11):** Các căn nhà mới hơn (35 tuổi trở xuống) có giá trị trung vị (MEDV) cao nhất. Khi tuổi thọ của căn nhà tăng lên (đặc biệt nhóm >=70 tuổi), giá trị căn nhà có xu hướng giảm rõ rệt.

### Câu 13 & 14: Kiểm định Levene về phương sai
Sử dụng thư viện `scipy.stats` (hàm `stats.levene`). Mục đích nhằm kiểm tra giả định đồng nhất phương sai.

```python
group1 = boston_df[boston_df['age_group'] == '35 tuổi trở xuống']['MEDV']
group2 = boston_df[boston_df['age_group'] == 'Từ 35 đến 70 tuổi']['MEDV']
group3 = boston_df[boston_df['age_group'] == '70 tuổi trở lên']['MEDV']

levene_stat, levene_p = stats.levene(group1, group2, group3)
print(f'Levene Statistic: {levene_stat:.4f}, p-value: {levene_p:.4f}')
```

### Câu 15: Kiểm định T-Test
Kiểm định xem có sự khác biệt về trung vị giá nhà (MEDV) giữa các nhà giáp sông Charles và không giáp sông.

```python
chas_0 = boston_df[boston_df['CHAS'] == 0]['MEDV']
chas_1 = boston_df[boston_df['CHAS'] == 1]['MEDV']
t_stat, t_p = stats.ttest_ind(chas_0, chas_1)
print(f'T-statistic: {t_stat:.4f}, p-value: {t_p:.4f}')
```

### Câu 16: Phép phân tích ANOVA và giả thuyết
Giả thuyết $H_0$: Trung bình giá nhà `MEDV` giữa các nhóm tuổi `AGE` là như nhau.
Giả thuyết $H_1$: Có ít nhất một nhóm tuổi có trung bình giá nhà khác biệt.

```python
f_stat, p_val = stats.f_oneway(group1, group2, group3)
print(f'ANOVA F-statistic: {f_stat:.4f}, p-value: {p_val:.4f}')
```

### Câu 17: Hệ số tương quan Pearson (r)
Đo lường mức độ và chiều hướng mối quan hệ tuyến tính giữa biến NOX và INDUS.

```python
pearson_r, pearson_p = stats.pearsonr(boston_df['NOX'], boston_df['INDUS'])
print(f'Pearson r: {pearson_r:.4f}, p-value: {pearson_p:.4f}')
```

### Câu 18: Hệ số góc trong Hồi quy tuyến tính
Xây dựng mô hình hồi quy tuyến tính đơn biến giữa giá trị nhà (MEDV) và tỷ lệ khoảng cách đến các trung tâm việc làm (DIS).

```python
X = boston_df['DIS']
y = boston_df['MEDV']
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
print(model.summary())
```

**Giải thích (Câu 18):** Hệ số góc ($\beta_1$) đại diện cho sự thay đổi trung bình của biến phụ thuộc `MEDV` khi khoảng cách `DIS` tăng thêm 1 đơn vị.
